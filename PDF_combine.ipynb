```python
from PyPDF2 import PdfMerger

# 要合併的 PDF 檔案
pdf_files = [
    "校園招募履歷表.pdf",
    "校園招募考生切結書.pdf",
    "國立高雄科技大學 在學證明書.pdf",
    "學士學位證書.pdf",
    "大學部歷年成績單_有章.pdf",
    "TOEIC 630.pdf",
    "專題.pdf",
    "相關競賽.pdf"

]

# 建立合併器
merger = PdfMerger()

# 依序加入 PDF
for pdf in pdf_files:
    merger.append(pdf)

# 輸出合併後的 PDF
output_file = "文件合併.pdf"
merger.write(output_file)
merger.close()

print(f"PDF 已成功合併成：{output_file}")
```
